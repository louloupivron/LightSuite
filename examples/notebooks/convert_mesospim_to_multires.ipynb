# Convert mesoSPIM exports to LightSuite Multires Pair v1

This notebook converts **mesoSPIM** overview / ROI hyperstacks (`*.tif` + `*_meta.txt`) into the canonical contract consumed by `lightsuite multires`.

| Output | Purpose |
|--------|---------|
| `multires_pair_manifest.json` | Physical frame + volume paths |

**After conversion:**

```bash
uv sync --extra registration
uv run lightsuite multires validate-config -c examples/config/multiresolution/JulieBuron_multires_manifest.yaml
uv run lightsuite multires check-geometry      -c ...
uv run lightsuite multires register            -c ...
```

````python
from pathlib import Path

OVERVIEW_TIFF = Path("/media/gbm/NVME2/ALICe-pipelines-data/JulieBuron/1X/1-561-1x.tif")
ROI_TIFF = Path("/media/gbm/NVME2/ALICe-pipelines-data/JulieBuron/3.2X/1-561-3.2x_TC.tif")

SAMPLE_NAME = "JulieBuron"
PAIR_LABEL = "baseline_561_TC"
SAVE_PATH = Path("/media/gbm/NVME2/ALICe-pipelines-data/JulieBuron/multiresolution_results")
CONVERTED_DIR = SAVE_PATH / "converted"
CONVERTED_DIR.mkdir(parents=True, exist_ok=True)
MANIFEST_PATH = CONVERTED_DIR / f"{PAIR_LABEL}_pair.json"
print(MANIFEST_PATH)
````

````python
from lightsuite.multires.vendor.mesospim import build_mesospim_pair_manifest

manifest = build_mesospim_pair_manifest(
    sample_name=SAMPLE_NAME,
    pair_label=PAIR_LABEL,
    overview_path=OVERVIEW_TIFF,
    roi_path=ROI_TIFF,
    output_manifest_path=MANIFEST_PATH,
)
print(manifest.overview.spacing_um)
print(manifest.overview.origin_um)
print("Wrote", MANIFEST_PATH)
````
